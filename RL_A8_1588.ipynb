State Value Function and Action Value Function

CALCULATING FOR STATE VALUE

```python
gamma = float(input("Enter the discount factor (gamma): "))

policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}


print("\nEnter rewards for the policy trajectory:")

rewards = []
states = ["S0", "S1", "S3", "S4"]

for state in states:
    r = float(input(f"Enter reward for {state}: "))
    rewards.append(r)


state_value = 0

for t in range(len(rewards)):
    state_value += (gamma ** t) * rewards[t]

print("\nEntered reward values:", rewards)
print("State Value V(S0):", round(state_value, 3))




```

CALCULATING FOR ACTION VALUE

```python
print("\nCalculate action value for S0")

action = input("Enter action at S0 (up/down/left/right): ").lower()

if action == "up":
    first_reward = float(input("Enter reward for S0 taking Up: "))

    action_rewards = [first_reward] + rewards

    action_value = 0
    for t in range(len(action_rewards)):
        action_value += (gamma ** t) * action_rewards[t]

    print("Action reward values:", action_rewards)
    print("Action Value Q(S0, Up):", round(action_value, 3))

else:
    print("This example calculates the extra first step for Up only.")
```
