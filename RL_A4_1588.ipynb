```python
import math
machines = ['A', 'B', 'C']
N_play = 10
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}
def calculate_UCB(t):

    for machine in machines:
        if N[machine] == 0:
            UCB[machine] = float('inf')

        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(t)) / N[machine]
            )

def select_machine():

    return max(machines, key=lambda x: UCB[x])


print("==============================================")
print(" UPPER CONFIDENCE BOUND - MULTI-ARMED BANDIT")
print("==============================================")

for play in range(1, N_play + 1):
    calculate_UCB(play)
    selected_machine = select_machine()

    print("\n----------------------------------------------")
    print("Play:", play)

    print("\nUCB Values:")

    for machine in machines:

        if UCB[machine] == float('inf'):
            print(machine, "=", "Infinity")
        else:
            print(machine, "=", round(UCB[machine], 4))

    print("\nSelected Machine:", selected_machine)
    reward = float(
        input("Enter reward obtained from Machine "
              + selected_machine + ": ")
    )
    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine])
        / N[selected_machine]
    )

    print("\nUpdated Values:")

    print("N =", N)
    print("R =", R)
    print("Q =", {
        m: round(Q[m], 4) for m in machines
    })
calculate_UCB(N_play)

total_reward = sum(R.values())

print("\n\n==========================================================")
print("                     FINAL RESULTS")
print("==========================================================")

print("\nMachine\tN\tR\t\tUCB\t\tQ")

print("----------------------------------------------------------")

for machine in machines:

    print(
        machine,
        "\t",
        N[machine],
        "\t",
        round(R[machine], 2),
        "\t\t",
        round(UCB[machine], 4),
        "\t\t",
        round(Q[machine], 4)
    )

print("----------------------------------------------------------")

print("\nTotal Reward Received =", total_reward)

print("\nFinal Q-values:")

for machine in machines:
    print(
        "Q(" + machine + ") =",
        round(Q[machine], 4)
    )
```
