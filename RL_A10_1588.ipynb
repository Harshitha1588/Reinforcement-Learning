Bellman Optimality and Value Iteration

```python


gamma = 0.9
theta = 0.000001

states = ["S0", "S1", "G"]
actions = ["up", "down", "left", "right"]


grid = [["S0", "S1"], ["W", "G"]]

positions = {"S0": (0, 0), "S1": (0, 1), "G": (1, 1)}

moves = {
    "up": (-1, 0),
    "down": (1, 0),
    "left": (0, -1),
    "right": (0, 1)
}


print("STATES:", states)
print("ACTIONS:", actions)
print("Discount Factor:", gamma)
print("Normal Reward: -1")
print("Wall Reward: -1")

def transition(state, action):
    r, c = positions[state]
    dr, dc = moves[action]

    nr = r + dr
    nc = c + dc

    if nr < 0 or nr >= 2 or nc < 0 or nc >= 2:
        return state, -1

    next_state = grid[nr][nc]

    if next_state == "W":
        return state, -1

    if next_state == "G":
        return "G", 10

    return next_state, -1


transitions = {}

print("\nTRANSITION VERIFICATION")

for s in ["S0", "S1"]:
    for a in actions:
        ns, reward = transition(s, a)
        transitions[(s, a)] = (ns, reward)
        print(s, a, "->", ns, "Reward:", reward)

V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

print("\nINITIAL VALUES:", V)


iteration = 0

print("\nVALUE ITERATION")
print("Iteration | V(S0) | V(S1) | Delta")

while True:
    newV = {"S0": 0.0, "S1": 0.0, "G": 0.0}

    for s in ["S0", "S1"]:
        q_values = []

        for a in actions:
            ns, reward = transitions[(s, a)]
            q = reward + gamma * V[ns]
            q_values.append(q)

        newV[s] = max(q_values)

    delta = max(
        abs(newV[s] - V[s]) for s in ["S0", "S1"]
    )

    iteration += 1
    V = newV

    print(iteration, "|", round(V["S0"], 6),
          "|", round(V["S1"], 6), "|", round(delta, 6))

    if delta < theta:
        break


print("\nFINAL OPTIMAL STATE VALUES")

for s in states:
    print("V*(", s, ") =", round(V[s], 6))


print("\nACTION VALUES AND OPTIMAL POLICY")

Q = {}

for s in ["S0", "S1"]:
    for a in actions:
        ns, reward = transitions[(s, a)]
        Q[(s, a)] = reward + gamma * V[ns]

    best_value = max(Q[(s, a)] for a in actions)

    best_actions = [
        a for a in actions
        if abs(Q[(s, a)] - best_value) < 1e-9
    ]

    print("\nState:", s)

    for a in actions:
        print("Q(", s, ",", a, ") =", round(Q[(s, a)], 6))

    print("Optimal action(s):", best_actions)


print("\nBELLMAN OPTIMALITY VERIFICATION")

for s in ["S0", "S1"]:
    max_q = max(Q[(s, a)] for a in actions)

    print(
        s,
        "V* =", round(V[s], 6),
        "max Q* =", round(max_q, 6),
        "Verified =", abs(V[s] - max_q) < 1e-9
    )

print("\nEXPLANATION")
print("S1: Moving down reaches the goal and gives reward +10.")
print("S0: Moving right reaches S1 and gives optimal value 8.")
print("The goal reward increases the values of states near the goal.")
print("The discount factor determines the importance of future rewards.")
```
